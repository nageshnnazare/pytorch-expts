```python
import torch
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score
```

# Simple Neural Network

```python
# Load dataset
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
print(df.head())
print(f"shape : {df.shape}")

print("-"*80)

# drop id / unnamed col
df.drop(columns=['id', 'Unnamed: 32'], inplace=True)
print(df.head())
print(f"shape : {df.shape}")
```

```python
# Preprocessing
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:],
                                                    df.iloc[:, 0],
                                                    test_size=0.2)
print(f"X_train shape : {X_train.shape}")
print(f"X_test shape : {X_test.shape}")
print(f"y_train shape : {y_train.shape}")
print(f"y_test shape : {y_test.shape}")

print("-"*80)

# Normalize input
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
print(f"X_train = \n{X_train}")
print(f"X_test = \n{X_test}")

print("-"*80)

# Relabel output
le = LabelEncoder()
y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)
print(f"y_train = \n{y_train}")
print(f"y_test = \n{y_test}")
```

```python
# Numpy to Pytorch handover
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)

print(f"X_train_tensor = \n{X_train_tensor}")
print(f"X_test_tensor = \n{X_test_tensor}")
print(f"y_train_tensor = \n{y_train_tensor}")
print(f"y_test_tensor = \n{y_test_tensor}")
```

```python
# Parameters
learning_rate = 0.1
epochs = 100
epsilon = 1e-6

# Model
class NeuralNet():
    def __init__(self, X):
        self.weights = torch.rand(X.shape[1], 1,
                                  dtype=torch.float64, requires_grad=True)
        self.bias = torch.zeros(1,
                                dtype=torch.float64, requires_grad=True)

    def forward(self, X):
        z = torch.matmul(X, self.weights) + self.bias
        y_pred = torch.sigmoid(z)
        return y_pred

    def loss(self, y_pred, y):
        y_pred = torch.clamp(y_pred, epsilon, 1-epsilon)
        loss = -y*torch.log(y_pred) - (1-y)*torch.log(1-y_pred)
        return loss.mean()
```

```python
# Create Model
model = NeuralNet(X_train_tensor)
print(f"model.weights = \n{model.weights}")
print(f"model.bias = \n{model.bias}")
```

```python
# Training loop
for epoch in range(epochs):
    # Forward pass
    y_pred = model.forward(X_train_tensor)

    # Loss computation
    loss = model.loss(y_pred, y_train_tensor)
    if epoch % 5 == 0:
      print(f"epoch : {epoch}, loss = {loss}")

    # Backprop
    loss.backward()

    # Update params
    with torch.no_grad():
        model.weights -= learning_rate * model.weights.grad
        model.bias -= learning_rate * model.bias.grad

    # clear accumulating grads
    model.weights.grad.zero_()
    model.bias.grad.zero_()
```

```python
# Evaluate model
with torch.no_grad():
    y_pred = model.forward(X_test_tensor)
    y_pred_class = (y_pred > 0.7).float()
    acc = (y_pred_class == y_test_tensor).float().mean()
    print(f"accuracy = {acc*100}%")
```

```python
import torch
import torch.nn as nn
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score
```

```python
# Load dataset
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')

# drop id / unnamed col
df.drop(columns=['id', 'Unnamed: 32'], inplace=True)

# Preprocessing
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:],
                                                    df.iloc[:, 0],
                                                    test_size=0.2)
# Normalize input
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Relabel output
le = LabelEncoder()
y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)

# Numpy to Pytorch handover
X_train_tensor = torch.from_numpy(X_train).float()
X_test_tensor = torch.from_numpy(X_test).float()
y_train_tensor = torch.from_numpy(y_train).float().unsqueeze(1)
y_test_tensor = torch.from_numpy(y_test).float().unsqueeze(1)
```

```python
# Parameters
learning_rate = 0.1
epochs = 100
epsilon = 1e-6

# Define Model
class NeuralNetwork(nn.Module):
    def __init__(self, num_features):
      super().__init__()
      self.network = nn.Sequential(
          nn.Linear(num_features, 1),
          nn.Sigmoid()
      )
      self.loss_func = nn.BCELoss()

    def forward(self, features):
      out = self.network(features)
      return out

    def loss(self, y_pred, y):
        return self.loss_func(y_pred, y)
```

```python
# Create Model
model = NeuralNetwork(X_train_tensor.shape[1])

# Training loop
for epoch in range(epochs):
    # Forward pass
    y_pred = model(X_train_tensor)

    # Loss computation
    loss = model.loss(y_pred, y_train_tensor)
    if epoch % 5 == 0:
      print(f"epoch : {epoch}, loss = {loss}")

    # Backprop
    loss.backward()

    # Update params
    with torch.no_grad():
        for param in model.parameters():
            param -= learning_rate * param.grad
            param.grad.zero_()
```

```python
# Evaluate model
with torch.no_grad():
    y_pred = model.forward(X_test_tensor)
    y_pred_class = (y_pred > 0.5).float()
    acc = (y_pred_class == y_test_tensor).float().mean()
    print(f"accuracy = {acc*100}%")
```
