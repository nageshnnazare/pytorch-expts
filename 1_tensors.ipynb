# Reduction operations

```python
t = torch.randint(size=(2,3), low=0, high=10)
t
```

```python
torch.sum(t)
```

```python
torch.sum(t, dim=0)
```

```python
torch.sum(t, dim=1)
```

```python
torch.mean(t, dtype=torch.float32)
```

```python
torch.mean(t, dim=0, dtype=torch.float32)
```

```python
torch.mean(t, dim=1, dtype=torch.float32)
```

```python
torch.median(t)
```

```python
torch.max(t)
```

```python
torch.min(t)
```

```python
torch.prod(t)
```

```python
torch.argmax(t)
```

```python
torch.argmin(t)
```

# Matrix operations

```python
f = torch.randint(size=(2,3), low=0, high=10)
g = torch.randint(size=(3,2), low=0, high=10)
print(f)
print(g)
```

```python
torch.matmul(f, g)
```

```python
v1 = torch.tensor([1, 2])
v2 = torch.tensor([3, 4])
torch.dot(v1, v2)
```

```python
t = torch.tensor([[1, 2],[3, 4]], dtype=float)
torch.transpose(t, 0, 1)
```

```python
torch.det(t)
```

```python
torch.inverse(t)
```
